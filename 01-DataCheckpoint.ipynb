## Research Question

 - Is there a statistically significant association between an individual's demographic profile (specifically, age group and gender) and their primary motivation for listening to music (e.g., relaxation, social gathering, sadness)?

We set the feature of ages and genders as independent variables, and music-listening motivation as a dependent variable. To determine if a relationship exists, we will approach this as a statistical inference task. Since our dependent and independent variables are all categorical, we plan to use Chi-square tests of independence.

## Background and Prior Work

 - Music functions in individuals’ daily lives as more than mere entertainment; it serves as an important tool for emotion regulation, the formation of self-identity, and the development of social connections. The reasons why people listen to music vary widely, including relaxation, alleviating loneliness, and gaining energy. Understanding how these motivations differ according to demographic attributes such as age and gender is an important issue in music psychology. This study aims to determine, using  statistical methods, whether specific age groups or genders have statistically significant associations with primary motivations for listening to music.

 - Previous research has shown that gender is a powerful factor influencing musical preferences and listening behavior. Colley (2008) reported that genre preferences among younger populations are closely tied to sex and gender-related personality traits. <a name="back1"> </a>[1](#ref1) In addition, age-related changes are an undeniable factor in musical engagement. A cross-sectional study by Vieillard & Gilet (2013) emphasized that older adults, compared to younger populations, are less likely to perceive negative emotions such as sadness or fear from music, highlighting that aging is associated with a relative preference for positive over negative emotions. <a name="back2"> </a>[2](#ref2) Recent industry data further suggests that for specific demographics, such as adult women, music increasingly serves a medicinal role in coping with life challenges. <a name="back4"> </a>[4](#ref4)

 - In this project, we set age and gender as independent variables and music listening motivation as the dependent variable. Given that these are categorical data, we will employ the Pearson’s Chi-square test of independence. As established by McHugh (2013), the Chi-square test is the most appropriate non-parametric tool for evaluating whether a statistically significant association exists between nominal variables, providing the necessary robustness for frequency-based demographic data.<a name="back3"> </a>[3](#ref3)

## Hypothesis


We hypothesize that music listening motivations will vary significantly across age and gender groups. Specifically we predict: 
- Adolescents (12-20): This group will primarily utilize music for social connection and emotional expression (sadness/relief). This prediction is supported by Rentfrow and Gosling (2003) <a name="back5"> </a>[5](#ref5), who found that younger populations utilize 'Reflective and Complex' music to process emotions and 'Upbeat' music for social gatherings.
- Young Adults (20–35): We predict this group will demonstrate a transitional motivation, focusing on both social gatherings and relaxation/stress relief. This aligns with Bonneville-Roussy et al. (2013) <a name="back6"> </a>[6](#ref6), who noted that while social music use remains high in young adulthood, there is an emerging shift toward using music for self-regulation as individuals enter the workforce.
- Established Adults and Seniors (35–60 and 60+): We hypothesize that these groups will primarily utilize music for relaxation and stress relief. According to Vieillard & Gilet (2013) [2](#ref2), aging is associated with a preference for positive emotional states and a reduction in music used for social identity, favoring music that serves as a tool for comfort and mood stability.
- Gender (Women 20+): We predict that women within the adult age bins (20–35 and 35+) will show a statistically higher association with relaxation than their male counterparts. This is based on Edison Research (2025) [4](#ref4), which reported that 58% of women specifically identify music as a primary tool for stress relief and 'medicine' for coping with life challenges.

### Dataset

  - Dataset name: Spotify User Behavior Survey Data
  - Link to the dataset: https://www.kaggle.com/datasets/coulsonlll/spotify-user-behavior-survey-data
  - Number of observations: 28,546 in the raw file; 27,543 after excluding the "6–12" age group during cleaning
  - Number of variables: 20 (raw); 6 retained after cleaning (Age, Gender, and four one-hot encoded mood columns)
  - Description of the variables most relevant to this project

    1. Age: A categorical variable indicating the respondent's age group, defining "12~20" as "Adolescents", "20~35" as "Young Adults", and both "35~60" and "60+" as "Established Adults and Seniors". This variable serves as an independent variable for Chi-square tests of independence.
    2. Gender: A categorical variable representing the respondent's gender including "Female", "Male", and "Others". This is also utilized as an independent variable.
    3. music_Influencial_mood: A categorical variable indicating the primary motivation or mood for listening to music including 4 categories: "Relaxation and stress relief", "Uplifting and motivational", "Sadness or melancholy", and "Social gatherings or parties". This serves as the dependent variable to analyze whether demographic profiles significantly influence listening motivations.

- Methodological note on the dependent variable

    Because respondents could select multiple motivations, `music_Influencial_mood` is one-hot encoded into four binary columns rather than treated as a single categorical column. As a result, the analysis will consist of **four separate Chi-square tests of independence** — one per motivation, against each demographic variable (Age and Gender) — rather than a single 3×4 contingency test. This preserves the independence-of-observations assumption that an `explode`-based approach would have violated.

- Descriptions of any shortcomings this dataset has with respect to the project
    
    1. Sampling Bias: The survey respondents are predominantly concentrated in specific age groups (particularly 12–35 years old) and existing Spotify users. Consequently, the sample may not objectively represent the general population.
    2. Self-reported Data: Since the dataset is based on subjective survey responses, it is susceptible to response bias or inaccuracies stemming from the respondents' personal perceptions and memory regarding their primary motivations.
    3. Age-related Data Reliability: The dataset contains respondents in the "6–12" age group. However, Spotify's terms of service<a name="back7"> </a>[7](#ref7) generally require users to be 13 or older (users under 13 typically require a Family Plan). This suggests that data for the "6–12" group may reflect shared family usage rather than individual preferences. To ensure data integrity, this specific age group was excluded from the analysis.
    4. Multiple choice responses: The "music_Influencial_mood" column contained multiple comma-separated responses per user. Initially, we considered using `explode` to unnest these values. However, doing so would create duplicate respondent rows, violating the independence of observations assumption required for Chi-square tests. To maintain this, we instead used One-Hot Encoding. This transformed each unique motivation into its own binary variable (0 or 1) while strictly preserving the assumption.

<a name="ref7"> </a>[^](#back7) Spotify. (2024) Spotify Free and Premium Support: Age requirements. https://www.spotify.com/us/safetyandprivacy/parental-guide

```python
## YOUR CODE TO LOAD/CLEAN/TIDY/WRANGLE THE DATA GOES HERE
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

file_path = 'data/00-raw/Spotify_user_research.xlsx'
df = pd.read_excel(file_path, sheet_name='Sheet1')

# 1. Extract 3 columns
df_clean = df[['Age', 'Gender', 'music_Influencial_mood']].copy()

# 2. Exclude '6~12' aged users (under Spotify's minimum age of service)
df_clean = df_clean[df_clean['Age'] != '6~12'].copy()

# 3. Rename and merge age groups
age_group = {
    '12~20': 'Adolescents',
    '20~35': 'Young Adults',
    '35~60': 'Established Adults and Seniors',
    '60+': 'Established Adults and Seniors'
}
df_clean['Age'] = df_clean['Age'].replace(age_group)

# 4. Set Age as an ordered categorical so plots and crosstabs come out in
#    developmental order rather than insertion/alphabetical order.
age_order = ['Adolescents', 'Young Adults', 'Established Adults and Seniors']
df_clean['Age'] = pd.Categorical(df_clean['Age'], categories=age_order, ordered=True)

# 5. One-hot encode the multi-select mood column. We use get_dummies rather
#    than explode to avoid duplicating respondent rows, which would violate
#    the independence-of-observations assumption for Chi-square tests.
mood_dummies = df_clean['music_Influencial_mood'].str.get_dummies(sep=', ')
df_clean = pd.concat([df_clean[['Age', 'Gender']].reset_index(drop=True),
                      mood_dummies.reset_index(drop=True)], axis=1)

# 6. Sanity check: drop any respondents whose mood row is all zeros
#    (i.e., a blank/whitespace original response that produced no dummies).
mood_cols = mood_dummies.columns.tolist()
all_zero = (df_clean[mood_cols].sum(axis=1) == 0)
print(f'Respondents with no mood selected (dropped): {int(all_zero.sum())}')
df_clean = df_clean.loc[~all_zero].reset_index(drop=True)
```
