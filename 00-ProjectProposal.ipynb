## Research Question

 - Is there a statistically significant association between an individual's demographic profile (specifically, age group and gender) and their primary motivation for listening to music (e.g., relaxation, social gathering, sadness)?

 - We set the feature of ages and genders as independent variables, and music-listening motivation as a dependent variable.

 - To determine if a relationship exists, we will approach this as a statistical inference task. Since our dependent and independent variables are all categorical, we plan to use Chi-square tests of independence.

## Background and Prior Work

 - Music functions in individuals’ daily lives as more than mere entertainment; it serves as an important tool for emotion regulation, the formation of self-identity, and the development of social connections. The reasons why people listen to music vary widely, including relaxation, alleviating loneliness, and gaining energy. Understanding how these motivations differ according to demographic attributes such as age and gender is an important issue in music psychology. This study aims to determine, using  statistical methods, whether specific age groups or genders have statistically significant associations with primary motivations for listening to music.

 - Previous research has shown that gender is a powerful factor influencing musical preferences and listening behavior. Colley (2008) reported that genre preferences among younger populations are closely tied to sex and gender-related personality traits. <a name="back1"> </a>[1](#ref1) On the other hand, age-related changes are also an undeniable factor in the context of music listening. A cross-sectional study by Vieillard & Gilet (2013) emphasized that older adults, compared to younger populations, are less likely to perceive negative emotions such as sadness or fear from music, highlighting that aging is associated with a relative preference for positive over negative emotions. <a name="back2"> </a>[2](#ref2)

 - In this project, we integrate the findings of these previous studies, setting age and gender as independent variables and music listening motivation as the dependent variable. Given that all variables in this study are categorical data, we will use the Chi-square test to examine statistical associations. The reason for this is that the Spotify Musical Data Analysis study by divya-gh demonstrated that the Chi-square test was the best statistical measure to test hypotheses involving categorical data, such as release decades. <a name="back3"> </a>[3](#ref3) Through this, we will quantitatively evaluate whether groups with specific attributes tend to select music for particular purposes.
